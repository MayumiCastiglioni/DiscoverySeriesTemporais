# Pilotos dos quatro modelos

SARIMAX, Holt-Winters, Random Forest e XGBoost são ajustados uma vez por base na mesma primeira origem da agenda canônica. O notebook verifica a igualdade das origens, mas não calcula MAE nem ranking final.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / 'data').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))

import pandas as pd
from discovery_series_temporais.model_pilots import run_all_model_pilots
from discovery_series_temporais.series import build_origin_schedule, load_candidate_target

print(f'Raiz do projeto: {ROOT}')
```

```python
pilot_tables = []
for base_id in range(1, 6):
    candidate = load_candidate_target(base_id, ROOT / 'data' / 'processed')
    pilots = run_all_model_pilots(candidate, build_origin_schedule(candidate))
    assert pilots['origin_id'].nunique() == 1
    assert set(pilots['model']) == {'sarimax', 'holt_winters', 'random_forest', 'xgboost'}
    pilot_tables.append(pilots)

all_pilots = pd.concat(pilot_tables, ignore_index=True)
display(all_pilots[['model', 'origin_id', 'training_rows', 'random_state']])

output = ROOT / 'reports' / 'generated' / 'model_pilots' / 'shared_origin_model_pilots.csv'
output.parent.mkdir(parents=True, exist_ok=True)
all_pilots.to_csv(output, index=False)
print(f'Pilotos salvos em: {output.relative_to(ROOT)}')
print('Os erros desta execução são apenas diagnósticos de runtime e não devem ser comparados como MAE final.')
```
